```python
import pandas as pd
import scipy.stats as stats
```

```python
path = r"pokemon_team_rocket_dataset.csv"
data = pd.read_csv(path)
```

```python
data
```

```python
data.groupby(["Profession"])["Number of Gym Badges"].count()
```

```python
GroupA = data[data["Profession"] == "Fisherman"]["Number of Migrations"]
GroupB = data[data["Profession"] == "Breeder"]["Number of Gym Badges"]
GroupC = data[data["Profession"] == "Nurse"]["Number of Gym Badges"]

# Perform the Kruskal-Wallis H Test
h_statistic, p_value = stats.kruskal(GroupA, GroupB, GroupC)
 
# Output the results
print(f"H Statistic: {h_statistic}")
print(f"P-Value: {p_value}")
 
# Evaluate the hypothesis
alpha = 0.05
if p_value < alpha:
    print("We reject the null hypothesis. There is a significant difference between the groups.")
else:
    print("We fail to reject the null hypothesis. There is no significant difference between the groups.")
```
